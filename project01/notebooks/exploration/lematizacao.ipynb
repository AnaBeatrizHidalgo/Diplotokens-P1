```python
import stanza

# Just in the first time you run this, the English model will be downloaded
stanza.download('en')
```

```python
# Pipeline with tokenization + POS + lematization
nlp = stanza.Pipeline('en', processors='tokenize,pos,lemma')

text = """The patient presented with severe dyspnea and elevated heart rate.
          The physician prescribed antibiotics and ordered further examinations."""

doc = nlp(text)

print("Token → Lema")
print("-" * 30)
for sentence in doc.sentences:
    for word in sentence.words:
        print(f"{word.text:<20} → {word.lemma}")
```

```python
# Pipeline just the lematization, with tokens already provided
nlp = stanza.Pipeline('en', processors='tokenize,pos,lemma',
                       tokenize_pretokenized=True)  # with tokens already provided

tokens = [["The", "patient", "presented", "with", "severe", "dyspnea",
           "and", "elevated", "heart", "rates", "prescribed", "antibiotics"]]

doc = nlp(tokens)

print("Token → Lema")
print("-" * 30)
for sentence in doc.sentences:
    for word in sentence.words:
        print(f"{word.text:<20} → {word.lemma}")
```
